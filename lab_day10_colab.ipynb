```python
"""Drive persistence helpers, also embedded in the standalone Colab notebook."""
import json
import errno
import re
import shutil
import time
from datetime import datetime, timezone
from pathlib import Path


CHECKPOINT_MARKER = "checkpoint_complete.json"


def retry_io(operation, attempts=3):
    """Retry transient mounted-Drive failures, without hiding persistent errors."""
    for attempt in range(attempts):
        try:
            return operation()
        except OSError as error:
            if error.errno in (errno.ENOSPC, errno.EDQUOT, errno.EACCES, errno.EPERM) or attempt + 1 == attempts:
                raise
            delay = 2 ** attempt
            print(f'I/O tạm thời thất bại; thử lại sau {delay}s ({attempt + 1}/{attempts}).')
            time.sleep(delay)


def atomic_json(path, value):
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)
    temporary = path.with_name(path.name + ".tmp")
    encoded = json.dumps(value, ensure_ascii=False, indent=2)
    def write():
        temporary.write_text(encoded, encoding="utf-8")
        temporary.replace(path)
    retry_io(write)


def seal_checkpoint(directory, run):
    """Commit a snapshot only after weights and training state have been written."""
    directory = Path(directory)
    files = {p.relative_to(directory).as_posix(): p.stat().st_size
             for p in directory.rglob("*") if p.is_file()
             and p.name != CHECKPOINT_MARKER and not p.name.endswith(".tmp")}
    atomic_json(directory / CHECKPOINT_MARKER, {
        "format": 2, "step": run["completed_steps"],
        "fingerprint": run["fingerprint"], "files": files,
    })


def valid_checkpoint(directory, fingerprint=None):
    directory = Path(directory)
    try:
        marker = json.loads((directory / CHECKPOINT_MARKER).read_text(encoding="utf-8"))
        files = marker["files"]
        if marker.get("format") != 2 or not isinstance(files, dict):
            return False
        required = {"run.json", "training_state.pt", "adapter/adapter_config.json"}
        if not required <= files.keys() or not any(
                name.startswith("adapter/") and name.endswith((".safetensors", ".bin"))
                for name in files):
            return False
        run = json.loads((directory / "run.json").read_text(encoding="utf-8"))
        if run["config"]["train_projector"] and "extra_trainable.pt" not in files:
            return False
        if marker["step"] != run["completed_steps"] or marker["fingerprint"] != run["fingerprint"]:
            return False
        if fingerprint is not None and marker["fingerprint"] != fingerprint:
            return False
        for name, size in files.items():
            path = (directory / name).resolve()
            path.relative_to(directory.resolve())
            if not path.is_file() or path.stat().st_size != size or size == 0:
                return False
        return True
    except (OSError, ValueError, TypeError, KeyError):
        return False


class DriveStore:
    """Copy completed files atomically; publish snapshot markers last."""

    def __init__(self, output_root, saved_root, enabled=True):
        self.output = Path(output_root).resolve()
        self.saved = Path(saved_root).resolve()
        self.enabled = enabled
        if self.output == self.saved or self.output in self.saved.parents or self.saved in self.output.parents:
            raise ValueError("Local output and Drive backup must be separate directories.")

    @staticmethod
    def _files(source):
        files = [source] if source.is_file() else [p for p in source.rglob("*") if p.is_file()]
        return sorted((p for p in files if not p.name.endswith((".tmp", ".uploading", ".partial"))),
                      key=lambda p: (p.name == CHECKPOINT_MARKER,
                                     p.name.endswith(("run.json", ".manifest.json")), str(p)))

    def backup(self, source):
        if not self.enabled:
            return
        return retry_io(lambda: self._backup_once(source))

    def _backup_once(self, source):
        source = Path(source).resolve()
        relative = source.relative_to(self.output)
        if not source.exists():
            raise FileNotFoundError(source)
        destination = self.saved / relative
        if (source / CHECKPOINT_MARKER).is_file():
            # Reusing an already-published snapshot must not remove its commit marker.
            if valid_checkpoint(destination) and (
                    (source / CHECKPOINT_MARKER).read_bytes() == (destination / CHECKPOINT_MARKER).read_bytes()
                    and (source / 'run.json').read_bytes() == (destination / 'run.json').read_bytes()):
                return
            (destination / CHECKPOINT_MARKER).unlink(missing_ok=True)
        for file in self._files(source):
            dest = destination if source.is_file() else destination / file.relative_to(source)
            dest.parent.mkdir(parents=True, exist_ok=True)
            if dest.exists() and dest.stat().st_size == file.stat().st_size and (
                    dest.stat().st_mtime_ns == file.stat().st_mtime_ns):
                continue
            temporary = dest.with_name(dest.name + ".uploading")
            def copy_file():
                shutil.copy2(file, temporary)
                temporary.replace(dest)
            copy_file()
        if (source / CHECKPOINT_MARKER).is_file() and not valid_checkpoint(destination):
            raise IOError(f"Drive checkpoint verification failed: {destination}")
        atomic_json(self.saved / "backup_status.json", {
            "last_success_utc": datetime.now(timezone.utc).isoformat(),
            "last_path": relative.as_posix(),
        })

    def restore(self):
        return retry_io(self._restore_once)

    def _restore_once(self):
        if not self.saved.is_dir():
            return 0
        snapshots = {p.parent: valid_checkpoint(p.parent)
                     for p in self.saved.rglob(CHECKPOINT_MARKER)}
        repair = {p: not valid_checkpoint(self.output / p.relative_to(self.saved))
                  for p, valid in snapshots.items() if valid}
        restored = 0
        for source in self._files(self.saved):
            if any(parent in snapshots and not snapshots[parent] for parent in source.parents):
                continue
            # An interrupted new snapshot may not have its marker yet.
            parents = [p for p in source.parents if p == self.saved or self.saved in p.parents]
            snapshot = next((p for p in parents if p.name == "final" or re.fullmatch(r"step-\d+", p.name)), None)
            if snapshot is not None and snapshot not in snapshots:
                record = snapshot / "run.json"
                try:
                    legacy = json.loads(record.read_text(encoding="utf-8")).get("checkpoint_format", 1) < 2
                except (OSError, ValueError):
                    legacy = False
                if not legacy:
                    continue
            destination = self.output / source.relative_to(self.saved)
            if not destination.exists() or (snapshot is not None and repair.get(snapshot, False)):
                destination.parent.mkdir(parents=True, exist_ok=True)
                temporary = destination.with_name(destination.name + ".tmp")
                def copy_file():
                    shutil.copy2(source, temporary)
                    temporary.replace(destination)
                retry_io(copy_file)
                restored += 1
        return restored

    def prune_snapshots(self, directory, keep=1):
        """Remove only older committed step snapshots inside this exact run."""
        directory = Path(directory).resolve()
        relative = directory.relative_to(self.output)
        snapshots = sorted((p for p in directory.glob("step-*")
                            if re.fullmatch(r"step-\d+", p.name) and valid_checkpoint(p)),
                           key=lambda p: int(p.name.split("-")[1]))
        for old in snapshots[:-keep]:
            for base in (directory, self.saved / relative) if self.enabled else (directory,):
                target = base / old.name
                # Verify absolute target and parent before any recursive delete.
                if target.is_symlink() or target.resolve().parent != base.resolve():
                    raise ValueError(f"Unsafe checkpoint path: {target}")
                if target.exists():
                    try:
                        retry_io(lambda: shutil.rmtree(target))
                    except OSError:
                        # Retention cleanup is optional; the new checkpoint is already safe.
                        print(f'Chưa dọn được checkpoint cũ, giữ lại để không ngắt train: {target}')
```

```python
import math

def yolo_to_box(label, yolo):
    assert label in CLASSES and len(yolo) == 4, (label, yolo)
    assert all(math.isfinite(v) and 0 <= v <= 1 for v in yolo), (label, yolo)
    xc, yc, w, h = yolo
    box = [int(round(max(0, min(1000, value)))) for value in (
        (xc - w / 2) * 1000, (yc - h / 2) * 1000,
        (xc + w / 2) * 1000, (yc + h / 2) * 1000)]
    for lo, hi in ((0, 2), (1, 3)):
        if box[hi] <= box[lo]:
            box[lo] = min(box[lo], 999)
            box[hi] = box[lo] + 1
    return {'label': label, 'bbox_2d': box}


def objects_to_answer(objects):
    boxes = [yolo_to_box(obj['label'], obj['yolo']) for obj in objects]
    boxes.sort(key=lambda box: (box['bbox_2d'][1], box['bbox_2d'][0]))
    answer = json.dumps(boxes, separators=(',', ':'))
    assert parse_output(answer) is not None, 'Đáp án train không đúng JSON grounding.'
    return answer



PROMPT = (
    'Locate all potholes, cracks, and manholes in this image. '
    'Return only a JSON list of objects with keys label and bbox_2d. '
    'Labels must be pothole, crack, or manhole. '
    'bbox_2d is [x1,y1,x2,y2] using integer coordinates from 0 to 1000 '
    'relative to image width and height. Return [] if no objects are present.'
)
```

Tham khảo [Transformers 5.19 module matching](https://github.com/huggingface/transformers/blob/v5.19.0/src/transformers/quantizers/quantizers_utils.py).


```python
"""Download the pinned Qwen snapshot once; all subsequent model loads are local."""
import json
import re
import time
from pathlib import Path


def retry_download(operation, attempts=5):
    for attempt in range(attempts):
        try:
            return operation()
        except Exception as error:
            chain, transient = error, False
            seen = set()
            while chain is not None and id(chain) not in seen:
                seen.add(id(chain))
                status = getattr(getattr(chain, 'response', None), 'status_code', None)
                transient |= status in (408, 429, 500, 502, 503, 504)
                transient |= isinstance(chain, (TimeoutError, ConnectionError))
                transient |= type(chain).__name__ in (
                    'ConnectError', 'ReadError', 'TimeoutException', 'ConnectTimeout',
                    'ReadTimeout', 'WriteTimeout', 'RemoteProtocolError')
                chain = chain.__cause__ or chain.__context__
            if not transient or attempt + 1 == attempts:
                raise
            delay = min(2 ** (attempt + 1), 30)
            print(f'Hugging Face tạm thời không truy cập được; thử lại sau {delay}s.')
            time.sleep(delay)


def prepare_model_snapshot(model_id, revision, cache_dir, token=None):
    from huggingface_hub import hf_hub_download, try_to_load_from_cache

    def fetch(filename, pinned_revision):
        if re.fullmatch(r'[0-9a-f]{40}', pinned_revision):
            cached = try_to_load_from_cache(model_id, filename, revision=pinned_revision, cache_dir=cache_dir)
            if isinstance(cached, str) and Path(cached).is_file() and Path(cached).stat().st_size > 0:
                return Path(cached)
        return Path(retry_download(lambda: hf_hub_download(
            model_id, filename, revision=pinned_revision, cache_dir=cache_dir, token=token)))

    config_path = fetch('config.json', revision)
    commit = config_path.parent.name
    if not re.fullmatch(r'[0-9a-f]{40}', commit):
        raise RuntimeError(f'Không xác định được model commit từ snapshot: {config_path.parent}')
    files = ('generation_config.json', 'preprocessor_config.json', 'video_preprocessor_config.json',
             'tokenizer_config.json', 'tokenizer.json', 'vocab.json', 'merges.txt',
             'chat_template.json', 'model.safetensors.index.json')
    for filename in files:
        fetch(filename, commit)
    index = json.loads((config_path.parent / 'model.safetensors.index.json').read_text(encoding='utf-8'))
    shards = sorted(set(index['weight_map'].values()))
    if not shards:
        raise RuntimeError('Weight index không có shard.')
    for shard in shards:
        if Path(shard).name != shard or not shard.endswith('.safetensors'):
            raise ValueError(f'Tên model shard không hợp lệ: {shard}')
        fetch(shard, commit)
    print('Model snapshot sẵn sàng để nạp offline:', commit)
    return config_path.parent, commit
```

```python
from transformers import (AutoConfig, AutoProcessor, BitsAndBytesConfig,
                          Qwen3VLForConditionalGeneration, set_seed)
from bitsandbytes.nn import Linear4bit

revision_path = OUTPUT_ROOT / 'model_revision.json'
if revision_path.exists():
    saved_revision = json.loads(revision_path.read_text())
    assert saved_revision['model_id'] == MODEL_ID, 'Đổi RUN_LABEL khi thay model.'
    assert MODEL_REVISION in ('main', saved_revision['commit']), 'Model revision khác lượt cũ.'
    MODEL_REVISION = saved_revision['commit']
MODEL_LOCAL_PATH, MODEL_COMMIT = prepare_model_snapshot(
    MODEL_ID, MODEL_REVISION, CACHE_ROOT, token=os.environ.get('HF_TOKEN'))
model_config = AutoConfig.from_pretrained(MODEL_LOCAL_PATH, local_files_only=True)
atomic_json(revision_path, {'model_id': MODEL_ID, 'commit': MODEL_COMMIT})
backup_path(revision_path)
processor = AutoProcessor.from_pretrained(MODEL_LOCAL_PATH, local_files_only=True)
processor.tokenizer.padding_side = 'left'
if processor.tokenizer.pad_token_id is None:
    processor.tokenizer.pad_token = processor.tokenizer.eos_token
print('Model commit:', MODEL_COMMIT)


def load_base_model():
    quantization = BitsAndBytesConfig(
        load_in_4bit=True, bnb_4bit_quant_type='nf4',
        bnb_4bit_compute_dtype=torch.float16, bnb_4bit_use_double_quant=True,
        llm_int8_skip_modules=['visual', 'model.visual', 'lm_head'])
    model = Qwen3VLForConditionalGeneration.from_pretrained(
        MODEL_LOCAL_PATH, local_files_only=True,
        quantization_config=quantization, dtype=torch.float16,
        device_map={'': 0}, attn_implementation='sdpa')
    bad = [n for n, m in model.model.visual.named_modules() if isinstance(m, Linear4bit)]
    assert not bad, f'ViT bị lượng tử hóa: {bad[:5]}'
    assert not isinstance(model.lm_head, Linear4bit)
    assert any(isinstance(m, Linear4bit) for m in model.model.language_model.modules())
    return model


def user_message(image, prompt=PROMPT):
    return {'role': 'user', 'content': [{'type': 'image', 'image': image},
                                        {'type': 'text', 'text': prompt}]}


def to_device(batch):
    return {key: value.to(DEVICE) for key, value in batch.items()}


def fingerprint(value):
    encoded = json.dumps(value, sort_keys=True, separators=(',', ':')).encode()
    return hashlib.sha256(encoded).hexdigest()


def config_fingerprint(config):
    return fingerprint({'config': config, 'model': MODEL_ID, 'revision': MODEL_COMMIT,
                        'dataset': DATASET_VERSION, 'prompt': PROMPT, 'mode': RUN_MODE, 'packages': PINNED,
                        'pipeline': PIPELINE_VERSION, 'implementation': COLAB_SOURCE_SHA256})


def run_directory(name):
    return WORK_ROOT / ('full' if RUN_MODE == 'full' else 'smoke') / name.lower()


def prediction_path(name, split):
    folder = PREDICTION_ROOT / RUN_MODE
    folder.mkdir(parents=True, exist_ok=True)
    return folder / f'{name.lower()}_{split}.jsonl'


def prediction_manifest(run, entries, split):
    return {'run_id': run['run_id'], 'fingerprint': run['fingerprint'],
            'split': split, 'max_new_tokens': MAX_NEW_TOKENS,
            'image_ids_sha256': fingerprint([e['image_id'] for e in entries]),
            'n_images': len(entries)}


def valid_prediction_file(path, entries, expected):
    path = Path(path)
    try:
        manifest = json.loads(path.with_suffix('.manifest.json').read_text())
        if any(manifest.get(k) != value for k, value in expected.items()):
            return False
        rows = read_jsonl(path)
        ids = [row['image_id'] for row in rows]
        return (len(ids) == len(entries) and len(set(ids)) == len(ids)
                and set(ids) == {e['image_id'] for e in entries}
                and all(isinstance(row.get('raw'), str) for row in rows))
    except (OSError, ValueError, KeyError, TypeError):
        return False


@torch.inference_mode()
def predict_jsonl(model, entries, out_path, manifest):
    """Resume a matching JSONL prefix and back up progress at batch boundaries."""
    import gc
    model.eval()
    model.gradient_checkpointing_disable()
    processor.tokenizer.padding_side = 'left'
    out_path = Path(out_path)
    progress = out_path.with_suffix('.progress.jsonl')
    manifest_path = out_path.with_suffix('.manifest.json')
    progress_manifest_path = out_path.with_suffix('.progress.manifest.json')
    previous_manifest = json.loads(progress_manifest_path.read_text()) if progress_manifest_path.exists() else None
    if (progress.exists() or previous_manifest is not None) and previous_manifest != manifest:
        raise RuntimeError(f'{progress}: inference config đã đổi. Chọn RUN_LABEL mới.')
    expected_ids = [e['image_id'] for e in entries]
    completed = []
    if progress.exists():
        lines = progress.read_bytes().splitlines()
        for index, line in enumerate(lines):
            try:
                row = json.loads(line.decode('utf-8'))
            except (ValueError, UnicodeDecodeError):
                if index == len(lines) - 1:
                    break  # A disconnect can leave the last write incomplete.
                raise
            if index >= len(entries) or row.get('image_id') != expected_ids[index] or not isinstance(row.get('raw'), str):
                raise RuntimeError(f'{progress}: IDs không khớp thứ tự inference.')
            completed.append(row)
    progress.parent.mkdir(parents=True, exist_ok=True)
    temporary = progress.with_name(progress.name + '.tmp')
    temporary.write_text(''.join(json.dumps(row, ensure_ascii=False) + '\n' for row in completed), encoding='utf-8')
    temporary.replace(progress)
    atomic_json(progress_manifest_path, manifest)
    backup_path(progress_manifest_path)
    last_backup = time.perf_counter()
    batch_size = INFER_BATCH_SIZE

    def generate_batch(rows):
        messages = [[user_message(load_image(e))] for e in rows]
        inputs = processor.apply_chat_template(
            messages, tokenize=True, add_generation_prompt=True,
            return_dict=True, return_tensors='pt', padding=True)
        inputs.pop('token_type_ids', None)
        inputs = to_device(inputs)
        with torch.autocast('cuda', dtype=torch.float16):
            generated = model.generate(**inputs, do_sample=False, use_cache=True,
                                       max_new_tokens=MAX_NEW_TOKENS,
                                       pad_token_id=processor.tokenizer.pad_token_id)
        decoded = processor.batch_decode(generated[:, inputs['input_ids'].shape[1]:],
                                         skip_special_tokens=True, clean_up_tokenization_spaces=False)
        if len(decoded) != len(rows):
            raise RuntimeError('Số output generation không khớp số ảnh.')
        return decoded

    with progress.open('a', encoding='utf-8') as stream:
        try:
            start = len(completed)
            while start < len(entries):
                rows = entries[start:start + batch_size]
                retry = False
                try:
                    raw = generate_batch(rows)
                except torch.cuda.OutOfMemoryError:
                    if batch_size == 1:
                        raise
                    batch_size = max(1, batch_size // 2)
                    retry = True
                if retry:
                    # Leave the exception scope before freeing tensors retained by its traceback.
                    gc.collect()
                    torch.cuda.empty_cache()
                    print(f'Inference thiếu VRAM: giảm batch xuống {batch_size}, thử lại cùng ID.')
                    continue
                for entry, text in zip(rows, raw):
                    stream.write(json.dumps({'image_id': entry['image_id'], 'raw': text}, ensure_ascii=False) + '\n')
                stream.flush()
                now = time.perf_counter()
                if start == len(completed) or now - last_backup >= PREDICTION_BACKUP_SECONDS:
                    backup_path(progress)
                    last_backup = time.perf_counter()
                start += len(rows)
                print(f'Inference {manifest["split"]}: {start}/{len(entries)}')
        finally:
            stream.flush()
            backup_path(progress)
    progress.replace(out_path)
    atomic_json(manifest_path, manifest)
    assert valid_prediction_file(out_path, entries, manifest)
    backup_path(out_path)
    backup_path(manifest_path)


def score_predictions(path, name, run):
    assert RUN_MODE == 'full', 'Smoke không được tạo evidence test.'
    assert valid_prediction_file(path, test_samples, prediction_manifest(run, test_samples, 'test'))
    out = EVIDENCE_ROOT / f'{name.lower()}.json'
    meta = {key: run[key] for key in ('trainable_params', 'peak_vram_gb', 'train_minutes')}
    meta.update(run_id=run['run_id'], fingerprint=run['fingerprint'],
                base_setup=run['config']['setup'], train_fraction=run['config']['train_fraction'])
    if run['config']['setup'] == 'E0':
        meta['inference_minutes'] = run.get('inference_minutes', 0)
    command = [sys.executable, str(REPO_ROOT / 'scripts/score.py'),
               '--pred', str(path), '--splits-dir', str(DATA_ROOT / 'splits'),
               '--split', 'test', '--name', name, '--out', str(out)]
    for key, value in meta.items():
        command += ['--meta', f'{key}={value}']
    subprocess.run(command, check=True)
    backup_path(out)
    return json.loads(out.read_text())


def execute_e0():
    config = {'name': 'E0', 'setup': 'E0', 'train_fraction': 1.0, 'seed': 42}
    directory = run_directory('E0')
    directory.mkdir(parents=True, exist_ok=True)
    path = prediction_path('E0', 'test' if RUN_MODE == 'full' else 'val')
    rows = test_samples if RUN_MODE == 'full' else val_samples[:3]
    split = 'test' if RUN_MODE == 'full' else 'val'
    record_path = directory / 'run.json'
    run = json.loads(record_path.read_text()) if record_path.exists() else None
    if run and run.get('fingerprint') != config_fingerprint(config):
        raise RuntimeError('E0 Output khác config/model/dataset. Chọn OUTPUT_ROOT mới.')
    if run and valid_prediction_file(path, rows, prediction_manifest(run, rows, split)):
        print('Dùng lại predictions E0 đã hoàn tất.')
    else:
        set_seed(42)
        model = load_base_model()
        try:
            print('Params:', model.num_parameters(),
                  'ViT gồm projector:', sum(p.numel() for p in model.model.visual.parameters()))
            run = run or {'config': config, 'fingerprint': config_fingerprint(config),
                   'run_id': uuid.uuid4().hex, 'dataset_version': DATASET_VERSION,
                   'model_id': MODEL_ID, 'model_revision': MODEL_COMMIT, 'prompt': PROMPT,
                   'trainable_params': 0, 'train_minutes': 0, 'completed_steps': 0,
                   'peak_vram_gb': 0, 'inference_minutes': 0}
            atomic_json(record_path, run)
            backup_path(record_path)
            torch.cuda.synchronize()
            torch.cuda.reset_peak_memory_stats()
            started = time.perf_counter()
            predict_jsonl(model, rows, path, prediction_manifest(run, rows, split))
            torch.cuda.synchronize()
            run.update(peak_vram_gb=torch.cuda.max_memory_allocated() / 1024**3,
                       inference_minutes=run.get('inference_minutes', 0) + (time.perf_counter() - started) / 60)
            record_path.write_text(json.dumps(run, indent=2))
            backup_path(record_path)
        finally:
            del model
            gc.collect()
            torch.cuda.empty_cache()
    for row in read_jsonl(path)[:3]:
        print(row['image_id'], row['raw'])
    return score_predictions(path, 'E0', run) if RUN_MODE == 'full' else None


# E0 chạy sau preflight GPU ở phần 6, trước các lượt train chính thức.
```

## 4. Dữ liệu huấn luyện và mask loss


Chỉ JSON đáp án và token kết thúc được tính loss. Prompt, token ảnh và padding có labels=-100.
Kiểm tra chat-template prefix, 220 token ảnh và padding trước khi train.


```python
from torch.utils.data import Dataset, DataLoader
from torch.nn.utils.rnn import pad_sequence


def encode_example(image, prompt, answer, processor):
    user = user_message(image, prompt)
    assistant = {'role': 'assistant', 'content': [{'type': 'text', 'text': answer}]}
    prompt_enc = processor.apply_chat_template(
        [user], tokenize=True, add_generation_prompt=True,
        return_dict=True, return_tensors='pt', padding=False)
    full_enc = processor.apply_chat_template(
        [user, assistant], tokenize=True, add_generation_prompt=False,
        return_dict=True, return_tensors='pt', padding=False)
    n_prompt = prompt_enc['input_ids'].shape[1]
    assert torch.equal(full_enc['input_ids'][0, :n_prompt], prompt_enc['input_ids'][0]),         'Prompt không phải tiền tố: kiểm tra chat template.'
    end_id = processor.tokenizer.convert_tokens_to_ids('<|im_end|>')
    endings = (full_enc['input_ids'][0, n_prompt:] == end_id).nonzero().flatten()
    assert len(endings) == 1, 'Phải có đúng một EOS của assistant.'
    stop = n_prompt + int(endings[0]) + 1
    sample = {k: full_enc[k][0, :stop] for k in ('input_ids', 'attention_mask')}
    if 'mm_token_type_ids' in full_enc:
        sample['mm_token_type_ids'] = full_enc['mm_token_type_ids'][0, :stop]
    sample['labels'] = sample['input_ids'].clone()
    sample['labels'][:n_prompt] = -100
    sample['pixel_values'] = full_enc['pixel_values']
    sample['image_grid_thw'] = full_enc['image_grid_thw']
    assert (sample['labels'] != -100).any(), 'Không có token đáp án.'
    return sample


def collate_fn(batch, pad_token_id=None):
    pad_token_id = processor.tokenizer.pad_token_id if pad_token_id is None else pad_token_id
    result = {
        'input_ids': pad_sequence([b['input_ids'] for b in batch], batch_first=True,
                                  padding_value=pad_token_id),
        'attention_mask': pad_sequence([b['attention_mask'] for b in batch],
                                       batch_first=True, padding_value=0),
        'labels': pad_sequence([b['labels'] for b in batch], batch_first=True,
                               padding_value=-100),
        'pixel_values': torch.cat([b['pixel_values'] for b in batch], dim=0),
        'image_grid_thw': torch.cat([b['image_grid_thw'] for b in batch], dim=0),
    }
    if 'mm_token_type_ids' in batch[0]:
        assert all('mm_token_type_ids' in b for b in batch)
        result['mm_token_type_ids'] = pad_sequence([b['mm_token_type_ids'] for b in batch],
                                                  batch_first=True, padding_value=0)
    assert (result['labels'][result['attention_mask'] == 0] == -100).all()
    return result


class RoadDataset(Dataset):
    def __init__(self, entries):
        self.entries = entries

    def __len__(self):
        return len(self.entries)

    def __getitem__(self, index):
        entry = self.entries[index]
        return encode_example(load_image(entry), PROMPT,
                              objects_to_answer(entry['objects']), processor)


# Kiểm tra loss mask và số token ảnh mà không cần nạp model.
sample = RoadDataset(train_samples)[0]
image_token_count = int((sample['input_ids'] == processor.image_token_id).sum())
grid_count = int(sample['image_grid_thw'].prod(dim=-1).sum()) // processor.image_processor.merge_size**2
assert image_token_count == grid_count == 220, (image_token_count, grid_count)
assert (sample['labels'][sample['input_ids'] == processor.image_token_id] == -100).all()
decoded_answer = processor.tokenizer.decode(
    sample['labels'][sample['labels'] != -100], skip_special_tokens=False)
assert decoded_answer.startswith(objects_to_answer(train_samples[0]['objects']))
assert decoded_answer == objects_to_answer(train_samples[0]['objects']) + '<|im_end|>'
print('Token ảnh:', image_token_count, '\nToken được tính loss:', decoded_answer)
batch_check = collate_fn([sample, RoadDataset(train_samples)[1]])
print({k: tuple(v.shape) for k, v in batch_check.items()})
del sample, batch_check
```

## 5. LoRA và đóng băng cho E1/E2/E3


E1: 252 module LoRA LLM. E2: thêm toàn bộ projector fp32. E3: thêm 96 module LoRA ViT.
Giữ tham số trainable fp32 cho GradScaler; forward autocast fp16 trên GPU 0.

```python
from peft import LoraConfig, PeftModel, get_peft_model


def is_projector(name):
    return '.visual.merger.' in name or '.visual.deepstack_merger_list.' in name


def lora_pattern(setup_name):
    assert setup_name in ('E1', 'E2', 'E3')
    llm = (r'model\.language_model\.layers\.\d+\.'
           r'(?:self_attn\.(?:q|k|v|o)_proj|mlp\.(?:gate|up|down)_proj)')
    vit = (r'model\.visual\.blocks\.\d+\.'
           r'(?:attn\.(?:qkv|proj)|mlp\.(?:linear_fc1|linear_fc2))')
    return f'(?:{llm}|{vit})' if setup_name == 'E3' else llm


def load_config(name):
    with (REPO_ROOT / 'configs' / f'{name.lower()}.yaml').open() as stream:
        config = yaml.safe_load(stream)
    config.update(setup=config['name'], train_fraction=1.0, coord='norm1000',
                  batch_size=TRAIN_BATCH_SIZE, grad_accum=GRAD_ACCUM,
                  max_steps=300 if RUN_MODE == 'full' else 2)
    return config

def apply_setup(model, setup_name, config=None):
    setup_name = setup_name.upper()
    config = config or load_config(setup_name)
    pattern = lora_pattern(setup_name)
    matched = [n for n, _ in model.named_modules() if re.fullmatch(pattern, n)]
    expected = 348 if setup_name == 'E3' else 252
    assert len(matched) == expected, f'Khớp {len(matched)}/{expected}: xem model.named_modules().'
    for param in model.parameters():
        param.requires_grad_(False)
    model.enable_input_require_grads()
    model = get_peft_model(model, LoraConfig(
        r=config['lora']['r'], lora_alpha=config['lora']['alpha'],
        lora_dropout=config['lora']['dropout'], target_modules=pattern,
        bias='none', task_type='CAUSAL_LM',
    ))
    for name, param in model.named_parameters():
        if config['train_projector'] and is_projector(name):
            param.requires_grad_(True)
        if param.requires_grad:
            param.data = param.data.to(torch.float32)
            assert 'lora_' in name or (config['train_projector'] and is_projector(name))
    projector_n = sum(p.numel() for n, p in model.named_parameters()
                      if p.requires_grad and is_projector(n))
    if setup_name in ('E2', 'E3'):
        assert 100_000_000 < projector_n < 120_000_000, projector_n
    else:
        assert projector_n == 0
    model.gradient_checkpointing_enable(gradient_checkpointing_kwargs={'use_reentrant': False})
    model.config.use_cache = False
    model.config.text_config.use_cache = False
    model.print_trainable_parameters()
    print('LoRA modules:', len(matched), 'Projector trainable:', projector_n)
    return model
```

Checkpoint lưu mỗi 25 bước hoặc khoảng 5 phút tại ranh giới optimizer update; giữ checkpoint bước gần nhất và final.
Lưu LoRA, projector, processor, optimizer/scheduler/scaler, RNG, vị trí DataLoader và run.json;
marker hoàn tất được ghi cuối rồi sao lưu Drive. Run all ở phiên mới tiếp tục checkpoint hợp lệ gần nhất.
Log cập nhật lên Drive mỗi 10 bước hoặc khoảng 60 giây; không phải checkpoint trọng số ở mỗi bước.

```python
"""Training cell source embedded by prepare_colab.py into lab_day10.ipynb.

Execute in the notebook namespace after data/model/LoRA cells. This file keeps
the standalone notebook's training implementation reviewable without outputs.
"""
import numpy as np
from transformers import get_cosine_schedule_with_warmup
from peft.utils.save_and_load import load_peft_weights, set_peft_model_state_dict


def select_training_entries(entries, fraction, seed):
    if not 0 < fraction <= 1:
        raise ValueError('train_fraction phải trong (0,1].')
    indices = random.Random(seed).sample(range(len(entries)), len(entries))
    return [entries[i] for i in indices[:max(1, round(len(entries) * fraction))]]


def load_completed_run(directory, config):
    directory = Path(directory)
    record_path = directory / 'run.json'
    if not record_path.exists():
        return None
    try:
        run = json.loads(record_path.read_text(encoding='utf-8'))
    except (OSError, ValueError):
        return None
    if not isinstance(run, dict):
        return None
    if run.get('checkpoint_format', 1) >= 2 and not valid_checkpoint(directory):
        return None
    if not run.get('fingerprint'):
        return None
    if run.get('fingerprint') != config_fingerprint(config):
        raise RuntimeError(f'{directory}: config/model/dataset đã thay đổi. Dùng RUN_LABEL mới.')
    if not run.get('complete') or run['completed_steps'] != config['max_steps']:
        return None
    if not (directory / 'adapter/adapter_config.json').exists():
        raise RuntimeError(f'{directory}: thiếu adapter_config.json.')
    if not any((directory / 'adapter').glob('*.safetensors')):
        raise RuntimeError(f'{directory}: thiếu trọng số adapter.')
    if config['train_projector'] and not (directory / 'extra_trainable.pt').exists():
        raise RuntimeError(f'{directory}: thiếu projector.')
    return run


def save_checkpoint(model, directory, run, history, training_state):
    directory = Path(directory)
    directory.mkdir(parents=True, exist_ok=True)
    (directory / CHECKPOINT_MARKER).unlink(missing_ok=True)
    model.save_pretrained(directory / 'adapter', save_embedding_layers=False)
    processor.save_pretrained(directory / 'processor')
    if run['config']['train_projector']:
        extra = {n: p.detach().float().cpu().contiguous()
                 for n, p in model.get_base_model().named_parameters() if is_projector(n)}
        assert extra
        temporary = directory / 'extra_trainable.pt.tmp'
        torch.save(extra, temporary)
        temporary.replace(directory / 'extra_trainable.pt')
    temporary = directory / 'training_state.pt.tmp'
    torch.save(training_state, temporary)
    temporary.replace(directory / 'training_state.pt')
    (directory / 'loss.jsonl').write_text(''.join(json.dumps(row) + '\n' for row in history), encoding='utf-8')
    shutil.copy2(directory.parent / 'train_image_ids.json', directory / 'train_image_ids.json')
    atomic_json(directory / 'run.json', run)
    seal_checkpoint(directory, run)
    backup_path(directory)
    STORE.prune_snapshots(directory.parent, keep=KEEP_STEP_CHECKPOINTS)
    print('Checkpoint đã lưu và kiểm tra:', directory)


def latest_training_checkpoint(directory, config):
    expected = config_fingerprint(config)
    candidates = sorted(Path(directory).glob('step-*'), reverse=True)
    for path in candidates:
        if not valid_checkpoint(path):
            continue
        run = json.loads((path / 'run.json').read_text(encoding='utf-8'))
        if run['fingerprint'] != expected:
            raise RuntimeError('Checkpoint khác cấu hình. Chọn RUN_LABEL mới.')
        return path, run
    return None, None


def train_experiment(config):
    config = copy.deepcopy(config)
    directory = run_directory(config['name'])
    final = directory / 'final'
    completed = load_completed_run(final, config)
    if completed:
        print('Dùng lại checkpoint hoàn tất:', final)
        backup_path(final)
        return final
    checkpoint, previous_run = latest_training_checkpoint(directory, config) if RESUME_TRAINING else (None, None)
    if checkpoint is None and directory.exists() and any(directory.iterdir()):
        if not RESTART_INCOMPLETE:
            raise RuntimeError(f'{directory}: không có checkpoint resume hợp lệ. Bật RESTART_INCOMPLETE để train lại.')
        # Do not overwrite an old run with a different identity.
        for path in directory.glob('step-*/run.json'):
            try:
                saved_run = json.loads(path.read_text(encoding='utf-8'))
            except (OSError, ValueError):
                continue
            if not isinstance(saved_run, dict):
                continue
            if saved_run.get('fingerprint') != config_fingerprint(config):
                raise RuntimeError('Run cũ khác cấu hình. Chọn RUN_LABEL mới.')
        print('Không có trạng thái resume hợp lệ: bắt đầu từ bước 0.')
    directory.mkdir(parents=True, exist_ok=True)
    set_seed(config['seed'])
    rows = select_training_entries(train_samples, config['train_fraction'], config['seed'])
    atomic_json(directory / 'train_image_ids.json', [e['image_id'] for e in rows])
    model = optimizer = scheduler = scaler = params = iterator = saved = batch = output = loss = None
    history = []
    step = skipped = consecutive_skips = batches_in_epoch = 0
    run_id = uuid.uuid4().hex
    previous_minutes = previous_peak = 0
    try:
        model = apply_setup(load_base_model(), config['setup'], config)
        if checkpoint:
            result = set_peft_model_state_dict(model, load_peft_weights(str(checkpoint / 'adapter'), device='cpu'))
            assert not result.unexpected_keys, result.unexpected_keys
            if config['train_projector']:
                extra = torch.load(checkpoint / 'extra_trainable.pt', map_location='cpu', weights_only=True)
                parameters = dict(model.get_base_model().named_parameters())
                assert set(extra) == {n for n in parameters if is_projector(n)}
                with torch.no_grad():
                    for name, value in extra.items():
                        assert parameters[name].shape == value.shape
                        parameters[name].copy_(value.to(parameters[name].device))
                del extra, parameters
        model.train()
        params = [p for p in model.parameters() if p.requires_grad]
        optimizer = torch.optim.AdamW(params, lr=float(config['lr']), weight_decay=0, foreach=False)
        scheduler = get_cosine_schedule_with_warmup(
            optimizer, num_warmup_steps=max(1, int(config['max_steps'] * 0.05)),
            num_training_steps=config['max_steps'])
        scaler = torch.amp.GradScaler('cuda', init_scale=1024)
        generator = torch.Generator().manual_seed(config['seed'])
        loader = DataLoader(RoadDataset(rows), batch_size=config['batch_size'], shuffle=True,
                            collate_fn=collate_fn, num_workers=0, generator=generator)
        if checkpoint:
            # Only load state from this run's own trusted checkpoint directory.
            saved = torch.load(checkpoint / 'training_state.pt', map_location='cpu', weights_only=False)
            optimizer.load_state_dict(saved['optimizer'])
            scheduler.load_state_dict(saved['scheduler'])
            scaler.load_state_dict(saved['scaler'])
            step, skipped = previous_run['completed_steps'], previous_run['skipped_steps']
            run_id = previous_run['run_id']
            previous_minutes, previous_peak = previous_run['train_minutes'], previous_run['peak_vram_gb']
            history = read_jsonl(checkpoint / 'loss.jsonl')
            batches_in_epoch = saved['batches_in_epoch']
            generator.set_state(saved['epoch_generator_state'])
        # Replay the current epoch's deterministic sampler, then restore dropout RNG.
        epoch_generator_state = generator.get_state()
        iterator = iter(loader)
        if saved:
            for _ in range(batches_in_epoch):
                next(iterator)
            random.setstate(saved['rng_python'])
            np.random.set_state(saved['rng_numpy'])
            torch.set_rng_state(saved['rng_torch'])
            torch.cuda.set_rng_state_all(saved['rng_cuda'])
            saved = None
            print('Resume', config['name'], 'từ bước', step)
        (directory / 'loss.jsonl').write_text(''.join(json.dumps(row) + '\n' for row in history), encoding='utf-8')
        optimizer.zero_grad(set_to_none=True)
        torch.cuda.synchronize()
        torch.cuda.reset_peak_memory_stats()
        started = last_checkpoint = last_log_backup = time.perf_counter()

        def record(complete):
            torch.cuda.synchronize()
            return {'config': config, 'fingerprint': config_fingerprint(config), 'checkpoint_format': 2,
                    'run_id': run_id, 'complete': complete, 'completed_steps': step,
                    'skipped_steps': skipped, 'train_images': len(rows),
                    'trainable_params': sum(p.numel() for p in params),
                    'peak_vram_gb': max(previous_peak, torch.cuda.max_memory_allocated() / 1024**3),
                    'train_minutes': previous_minutes + (time.perf_counter() - started) / 60,
                    'dataset_version': DATASET_VERSION, 'model_id': MODEL_ID,
                    'model_revision': MODEL_COMMIT, 'prompt': PROMPT}

        def snapshot(path, complete=False):
            save_checkpoint(model, path, record(complete), history, {
                'optimizer': optimizer.state_dict(), 'scheduler': scheduler.state_dict(),
                'scaler': scaler.state_dict(), 'rng_python': random.getstate(),
                'rng_numpy': np.random.get_state(), 'rng_torch': torch.get_rng_state(),
                'rng_cuda': torch.cuda.get_rng_state_all(),
                'epoch_generator_state': epoch_generator_state,
                'batches_in_epoch': batches_in_epoch,
            })

        while step < config['max_steps']:
            loss_sum = 0.0
            for _ in range(config['grad_accum']):
                try:
                    batch = next(iterator)
                except StopIteration:
                    epoch_generator_state = generator.get_state()
                    iterator = iter(loader)
                    batches_in_epoch = 0
                    batch = next(iterator)
                batches_in_epoch += 1
                batch = to_device(batch)
                with torch.autocast('cuda', dtype=torch.float16):
                    output = model(**batch)
                    loss = output.loss / config['grad_accum']
                if not torch.isfinite(loss).item():
                    raise FloatingPointError('Loss nan/inf: kiểm tra mask/dtype và giảm lr.')
                loss_sum += loss.detach().item()
                scaler.scale(loss).backward()
                batch = output = loss = None
            scaler.unscale_(optimizer)
            norm = torch.nn.utils.clip_grad_norm_(params, 1.0)
            used_lr = optimizer.param_groups[0]['lr']
            # A nonfinite norm can result from reduction overflow even if individual gradients are finite.
            if not torch.isfinite(norm).item():
                scaler.update(new_scale=scaler.get_scale() / 2)
                optimizer.zero_grad(set_to_none=True)
                skipped += 1
                consecutive_skips += 1
                if consecutive_skips >= 8:
                    raise FloatingPointError('8 bước gradient overflow liên tiếp.')
                continue
            scaler.step(optimizer)
            scaler.update()
            optimizer.zero_grad(set_to_none=True)
            consecutive_skips = 0
            scheduler.step()
            step += 1
            entry = {'step': step, 'loss': loss_sum, 'lr': used_lr,
                     'grad_norm': float(norm),
                     'minutes': previous_minutes + (time.perf_counter() - started) / 60}
            history.append(entry)
            with (directory / 'loss.jsonl').open('a', encoding='utf-8') as stream:
                stream.write(json.dumps(entry) + '\n')
            now = time.perf_counter()
            if step == 1 or step % 10 == 0:
                print(f'{config["name"]}: {step}/{config["max_steps"]}, loss={loss_sum:.4f}')
            if step == 1 or step % 10 == 0 or now - last_log_backup >= LOG_BACKUP_SECONDS:
                backup_path(directory / 'loss.jsonl')
                atomic_json(directory / 'progress.json', record(False))
                backup_path(directory / 'progress.json')
                last_log_backup = time.perf_counter()
            if step < config['max_steps'] and (
                    step % CHECKPOINT_EVERY_STEPS == 0 or now - last_checkpoint >= CHECKPOINT_EVERY_SECONDS):
                snapshot(directory / f'step-{step:04d}')
                last_checkpoint = time.perf_counter()
        snapshot(final, complete=True)
        return final
    finally:
        model = optimizer = scheduler = scaler = params = iterator = saved = batch = output = loss = None
        gc.collect()
        torch.cuda.empty_cache()
```

### Kiểm tra GPU trước khi chạy E0–E4
Xác minh tất cả ảnh và thử forward/backward/optimizer trên setup train nặng nhất,
dùng ảnh có đáp án dài nhất. Trọng số thử bị bỏ; các lượt chính vẫn bắt đầu độc lập.
Sau khi PASS mới chạy E0 và E1–E4. Inference tự giảm batch khi gặp CUDA OOM.


```python
"""GPU preflight cell: verify the largest training example before the long E0 run."""


def preflight_training():
    # Check every input image before any long GPU job, including images in later splits.
    for entries in splits.values():
        for entry in entries:
            with Image.open(DATA_ROOT / entry['image_path']) as image:
                image.load()
                assert image.size == (entry['width'], entry['height']), entry['image_id']
    setups = [name for name in ('E1', 'E2', 'E3') if name in RUNS]
    if 'E4' in RUNS:
        setups += ['E3'] if E4_BASE_SETUP == 'auto' else [E4_BASE_SETUP]
    if not setups:
        print('Preflight: dữ liệu ảnh hợp lệ; không có setup train được chọn.')
        return
    setup = max(setups, key=lambda name: int(name[1:]))
    config = load_config(setup)
    # Detect incompatible old outputs BEFORE doing E0 or allocating the model.
    for name in set(setups):
        candidate = load_config(name)
        load_completed_run(run_directory(name) / 'final', candidate)
        latest_training_checkpoint(run_directory(name), candidate)
    longest = max(train_samples, key=lambda entry: len(processor.tokenizer.encode(
        objects_to_answer(entry['objects']), add_special_tokens=False)))
    model = optimizer = scaler = batch = result = loss = params = None
    try:
        model = apply_setup(load_base_model(), setup, config)
        model.train()
        params = [param for param in model.parameters() if param.requires_grad]
        optimizer = torch.optim.AdamW(params, lr=float(config['lr']), weight_decay=0, foreach=False)
        scaler = torch.amp.GradScaler('cuda', init_scale=1024)
        sample = RoadDataset([longest])[0]
        batch = to_device(collate_fn([sample] * config['batch_size']))
        torch.cuda.reset_peak_memory_stats()
        succeeded = False
        # Permit normal GradScaler backoff, as in the real training loop.
        for _ in range(8):
            optimizer.zero_grad(set_to_none=True)
            with torch.autocast('cuda', dtype=torch.float16):
                result = model(**batch)
                loss = result.loss
            if not torch.isfinite(loss).item():
                raise FloatingPointError('Preflight: loss không hữu hạn.')
            scaler.scale(loss).backward()
            scaler.unscale_(optimizer)
            norm = torch.nn.utils.clip_grad_norm_(params, 1.0)
            if torch.isfinite(norm).item():
                scaler.step(optimizer)
                scaler.update()
                succeeded = True
                break
            scaler.update(new_scale=scaler.get_scale() / 2)
            result = loss = None
        if not succeeded:
            raise FloatingPointError('Preflight: gradient overflow sau 8 lần thử.')
        torch.cuda.synchronize()
        status = dict(setup=setup, image_id=longest['image_id'],
                      peak_vram_gb=torch.cuda.max_memory_allocated() / 1024**3,
                      model_revision=MODEL_COMMIT, implementation=COLAB_SOURCE_SHA256)
        atomic_json(OUTPUT_ROOT / 'preflight.json', status)
        backup_path(OUTPUT_ROOT / 'preflight.json')
        print('Preflight GPU PASS:', status)
    finally:
        model = optimizer = scaler = batch = result = loss = params = None
        gc.collect()
        torch.cuda.empty_cache()


preflight_training()
E0_RESULT = execute_e0() if 'E0' in RUNS else None
```

```python
def restore_checkpoint(directory, config):
    directory = Path(directory)
    run = load_completed_run(directory, config)
    assert run is not None, f'Checkpoint chưa hoàn tất: {directory}'
    saved_processor = AutoProcessor.from_pretrained(directory / 'processor')
    assert saved_processor.tokenizer.get_vocab() == processor.tokenizer.get_vocab()
    assert saved_processor.chat_template == processor.chat_template
    model = load_base_model()
    try:
        model = PeftModel.from_pretrained(model, directory / 'adapter', is_trainable=False)
        if config['train_projector']:
            state = torch.load(directory / 'extra_trainable.pt', map_location='cpu', weights_only=True)
            parameters = dict(model.get_base_model().named_parameters())
            assert set(state) == {n for n in parameters if is_projector(n)}
            with torch.no_grad():
                for name, value in state.items():
                    param = parameters[name]
                    assert value.shape == param.shape
                    param.data = param.data.float()
                    param.copy_(value.to(param.device))
                    assert torch.equal(param.detach().cpu(), value), name
            del parameters, state
        return model, run
    except Exception:
        del model
        gc.collect()
        torch.cuda.empty_cache()
        raise


def infer_checkpoint(directory, config, split):
    run = load_completed_run(directory, config)
    assert run is not None
    if split == 'test':
        assert RUN_MODE == 'full' and run['completed_steps'] == 300
        rows = test_samples
    else:
        rows = val_samples[:VAL_LIMIT] if RUN_MODE == 'full' else val_samples[:3]
    path = prediction_path(config['name'], split)
    manifest = prediction_manifest(run, rows, split)
    if valid_prediction_file(path, rows, manifest):
        print('Dùng lại predictions:', path)
    else:
        model, run = restore_checkpoint(directory, config)
        try:
            predict_jsonl(model, rows, path, manifest)
        finally:
            del model
            gc.collect()
            torch.cuda.empty_cache()
    if split == 'test':
        return score_predictions(path, config['name'], run)
    predictions = {r['image_id']: r['raw'] for r in read_jsonl(path)}
    result = evaluate(rows, predictions, n_boot=200)
    record = dict(result, fingerprint=run['fingerprint'], run_id=run['run_id'],
                  dataset_version=DATASET_VERSION, split='val', mode=RUN_MODE)
    path.with_suffix('.metrics.json').write_text(json.dumps(record, indent=2))
    backup_path(path.with_suffix('.metrics.json'))
    print(config['name'], 'val ảnh:', len(rows), 'F1:', result['micro']['f1'],
          'JSON valid:', result['parse_valid_rate'])
    for row in read_jsonl(path)[:3]:
        print(row['image_id'], row['raw'])
    return record


def execute_setup(name, config=None):
    config = config or load_config(name)
    checkpoint = train_experiment(config)
    validation = infer_checkpoint(checkpoint, config, 'val')
    test = infer_checkpoint(checkpoint, config, 'test') if RUN_MODE == 'full' else None
    return {'checkpoint': str(checkpoint), 'val': validation, 'test': test}


RESULTS = {}
```

### Khôi phục baseline cũ trước E4
Cell này lấy checkpoint final, predictions và evidence E1--E3 còn thiếu từ `RECOVERY_SOURCE_ROOT`.
Để `None` dùng Drive của `RUN_LABEL` hiện tại; chỉ đặt đường dẫn run cũ khi cần.

```python
"""Restore completed E1--E3 artifacts from an older Drive run before E4.

This file is embedded as one Colab cell by ``prepare_colab.py``.  It only
copies missing files, so a current run always takes priority over an older one.
"""
from pathlib import Path
import shutil


def _copy_missing(source, destination):
    """Copy a file or directory tree without replacing an existing artifact."""
    source, destination = Path(source), Path(destination)
    if source.is_file():
        files = [(source, destination)]
    elif source.is_dir():
        files = [(path, destination / path.relative_to(source))
                 for path in sorted(source.rglob('*')) if path.is_file()]
    else:
        return 0
    copied = 0
    for old, new in files:
        if old.name.endswith(('.tmp', '.partial', '.uploading')) or new.exists():
            continue
        new.parent.mkdir(parents=True, exist_ok=True)
        temporary = new.with_name(new.name + '.recovering')
        shutil.copy2(old, temporary)
        temporary.replace(new)
        copied += 1
    return copied


def restore_completed_baselines(source_root):
    """Restore only the final E1--E3 artifacts that have test evidence."""
    source_root = Path(source_root).expanduser()
    if not source_root.is_dir():
        raise FileNotFoundError(f'Không thấy run cũ: {source_root}')
    copied = 0
    recovered = []
    mode = 'full' if RUN_MODE == 'full' else 'smoke'
    for name in ('E1', 'E2', 'E3'):
        key = name.lower()
        source_evidence = source_root / 'report' / 'evidence' / f'{key}.json'
        if not source_evidence.is_file():
            print(f'{name}: run cũ không có evidence, không khôi phục.')
            continue
        artifacts = (
            (source_root / 'work' / mode / key / 'final', run_directory(name) / 'final'),
            (source_root / 'predictions' / RUN_MODE / f'{key}_val.jsonl', prediction_path(name, 'val')),
            (source_root / 'predictions' / RUN_MODE / f'{key}_val.manifest.json',
             prediction_path(name, 'val').with_suffix('.manifest.json')),
            (source_root / 'predictions' / RUN_MODE / f'{key}_val.metrics.json',
             prediction_path(name, 'val').with_suffix('.metrics.json')),
            (source_root / 'predictions' / RUN_MODE / f'{key}_test.jsonl', prediction_path(name, 'test')),
            (source_root / 'predictions' / RUN_MODE / f'{key}_test.manifest.json',
             prediction_path(name, 'test').with_suffix('.manifest.json')),
            (source_evidence, EVIDENCE_ROOT / f'{key}.json'),
        )
        for source, destination in artifacts:
            copied += _copy_missing(source, destination)
        run = load_completed_run(run_directory(name) / 'final', load_config(name))
        if run is None:
            raise RuntimeError(
                f'{name}: evidence có mặt nhưng checkpoint final không đầy đủ hoặc không khớp cấu hình. '
                'Dùng RUN_LABEL/cấu hình của run cũ, hoặc chọn run cũ cùng phiên bản notebook.')
        recovered.append(name)
    print(f'Khôi phục trước E4 từ {source_root}: {copied} file; baseline hợp lệ: {recovered or "không có"}.')
    return recovered


# None: dùng SAVED_ROOT, vốn đã được khôi phục tự động ở đầu notebook.
# Muốn dùng run có nhãn khác, đổi thành Path('/content/drive/MyDrive/lab_day10_colab_runs/<run_cu>').
RECOVERY_SOURCE_ROOT = SAVED_ROOT if RECOVERY_SOURCE_ROOT is None else Path(RECOVERY_SOURCE_ROOT)
if 'E4' in RUNS:
    restore_completed_baselines(RECOVERY_SOURCE_ROOT)
else:
    print('Không chạy E4 nên không cần khôi phục baseline cũ.')
```

## 8. E4, bảng kết quả và bài nộp


E4 đổi **một biến: data_size 10%/100%**. Giá trị 100% dùng lại run gốc, không infer test lần nữa.
`auto` chọn setup có F1 cao nhất trên cùng tập val preview (mặc định 40 ảnh), hòa thì ưu tiên
VRAM thấp hơn. Đây là lựa chọn dựa trên preview, chưa phải kết luận thống kê chắc chắn.
Có thể đặt E4_BASE_SETUP cụ thể sau khi xem kết quả val. Đọc CI95 trên test khi viết kết luận.


```python
def choose_e4_base():
    if E4_BASE_SETUP != 'auto':
        return E4_BASE_SETUP
    candidates = []
    for name in ('E1', 'E2', 'E3'):
        config = load_config(name)
        directory = run_directory(name) / 'final'
        run = load_completed_run(directory, config)
        metrics_path = prediction_path(name, 'val').with_suffix('.metrics.json')
        if run and metrics_path.exists():
            metrics = json.loads(metrics_path.read_text())
            expected_n = min(len(val_samples), VAL_LIMIT if RUN_MODE == 'full' else 3)
            expected_rows = val_samples[:expected_n]
            expected_manifest = prediction_manifest(run, expected_rows, 'val')
            if (metrics.get('fingerprint') == run['fingerprint']
                    and metrics.get('run_id') == run['run_id'] and metrics['n_images'] == expected_n
                    and valid_prediction_file(prediction_path(name, 'val'), expected_rows, expected_manifest)):
                candidates.append((metrics['micro']['f1'], -run['peak_vram_gb'], name))
    assert candidates, 'Cần chạy val E1/E2/E3 hoặc khôi phục run cũ từ Drive trước E4.'
    print('E4 candidates (val F1, -VRAM, setup):', candidates)
    return max(candidates)[2]


if 'E4' in RUNS:
    base_name = choose_e4_base()
    base_config = load_config(base_name)
    base_directory = run_directory(base_name) / 'final'
    base_run = load_completed_run(base_directory, base_config)
    assert base_run is not None
    if RUN_MODE == 'full':
        baseline_predictions = prediction_path(base_name, 'test')
        assert valid_prediction_file(baseline_predictions, test_samples,
                                     prediction_manifest(base_run, test_samples, 'test')), (
            'Cần hoàn tất test của baseline với cấu hình inference hiện tại trước E4.')
    small_config = copy.deepcopy(base_config)
    small_config.update(name='E4_data_size_0.1', train_fraction=0.1)
    RESULTS['E4_data_size_0.1'] = execute_setup('E4_data_size_0.1', small_config)
    if RUN_MODE == 'full':
        baseline_predictions = prediction_path(base_name, 'test')
        assert valid_prediction_file(baseline_predictions, test_samples,
                                     prediction_manifest(base_run, test_samples, 'test'))
        RESULTS['E4_data_size_1.0'] = score_predictions(
            baseline_predictions, 'E4_data_size_1.0', base_run)
    (OUTPUT_ROOT / f'e4_selection_{RUN_MODE}.json').write_text(
        json.dumps({'base_setup': base_name, 'variable': 'data_size', 'values': [0.1, 1.0],
                    'val_limit': VAL_LIMIT, 'base_fingerprint': base_run['fingerprint']}, indent=2))

if 'E4' in RUNS:
    backup_path(OUTPUT_ROOT / f'e4_selection_{RUN_MODE}.json')
```

Ảnh GT màu xanh, prediction màu đỏ. Viết báo cáo bằng số của bạn và chọn ít nhất 6 ảnh.


```python
def evidence_table():
    headers = '| Run | Params | VRAM GB | Phút | F1 | CI95 | FP/ảnh | pothole | crack | manhole |'
    lines = [headers, '|---|---:|---:|---:|---:|---|---:|---:|---:|---:|']
    for path in sorted(EVIDENCE_ROOT.glob('e*.json')):
        d = json.loads(path.read_text())
        assert d['dataset_version'] == DATASET_VERSION and d['split'] == 'test', (
            f'{path}: evidence khác dataset/split hiện tại.')
        m = d['meta']
        ci = d['f1_ci95']
        per = d['per_class']
        lines.append(
            f"| {d['name']} | {m.get('trainable_params', 0):,.0f} | "
            f"{m.get('peak_vram_gb', 0):.2f} | {(m.get('inference_minutes', 0) if d['name'] == 'E0' else m.get('train_minutes', 0)):.2f} | "
            f"{d['micro']['f1']:.3f} | [{ci[0]:.3f}, {ci[1]:.3f}] | "
            f"{d['fp_per_image']:.3f} | {per['pothole']['f1']:.3f} | "
            f"{per['crack']['f1']:.3f} | {per['manhole']['f1']:.3f} |")
    table = '\n'.join(lines)
    display(Markdown(table))
    (REPORT_ROOT / 'results_table.md').write_text(table, encoding='utf-8')
    backup_path(REPORT_ROOT / 'results_table.md')
    return table


def visualize_predictions(name):
    predictions = {r['image_id']: r['raw'] for r in read_jsonl(prediction_path(name, 'test'))}
    good, bad = [], []
    for entry in test_samples:
        metrics = evaluate([entry], predictions, n_boot=40)['micro']
        item = (entry, metrics)
        valid = parse_output(predictions.get(entry['image_id'], '')) is not None
        (good if valid and metrics['fp'] == metrics['fn'] == 0 else bad).append(item)
    chosen = good[:3] + bad[:3]
    selected_ids = {e['image_id'] for e, _ in chosen}
    chosen += [item for item in good + bad if item[0]['image_id'] not in selected_ids][:6 - len(chosen)]
    folder = REPORT_ROOT / 'figures'
    folder.mkdir(parents=True, exist_ok=True)
    for index, (entry, metrics) in enumerate(chosen):
        raw = predictions.get(entry['image_id'], '')
        parsed = parse_output(raw)
        fig, ax = plt.subplots(figsize=(10, 6))
        ax.imshow(load_image(entry))
        draw_boxes(ax, json.loads(objects_to_answer(entry['objects'])), entry['width'], entry['height'], 'lime')
        draw_boxes(ax, parsed or [], entry['width'], entry['height'], 'red', linestyle='--')
        ax.set_title(f"{name} · {entry['image_id']}\nGT xanh / Pred đỏ · "
                     f"TP={metrics['tp']} FP={metrics['fp']} FN={metrics['fn']} · "
                     f"JSON valid={parsed is not None}", fontsize=9)
        ax.axis('off')
        fig.tight_layout()
        fig.savefig(folder / f'{name.lower()}_{index + 1:02d}.png', dpi=140)
        plt.show()
        plt.close(fig)
    backup_path(folder)
```

## 9. E5 tùy chọn — reward IoU

Phần bắt buộc E0–E4 đã có mã chạy. Bonus này giữ hàm reward/test CPU;
notebook chưa chạy GRPO hoặc tạo evidence E5.


```python
from scripts.score import iou_xyxy, parse_output


def compute_reward(completion: str, gt_boxes: list[dict]) -> float:
  pred = parse_output(completion)
  if pred is None:
    return 0.0  # JSON hỏng
  if not gt_boxes and not pred:
    return 1.0  # Cả hai cùng rỗng

  total_iou = 0.0
  all_labels = {g["label"] for g in gt_boxes} | {p["label"] for p in pred}
  for lab in all_labels:
    g = [x["bbox"] for x in gt_boxes if x["label"] == lab]
    p = [x["bbox"] for x in pred if x["label"] == lab]
    pairs = sorted(
        (
            (iou_xyxy(a, b), i, j)
            for i, a in enumerate(g)
            for j, b in enumerate(p)
        ),
        reverse=True,
    )
    used_g, used_p = set(), set()
    for v, i, j in pairs:
      if v < 0.5:
        break
      if i in used_g or j in used_p:
        continue
      used_g.add(i)
      used_p.add(j)
      total_iou += v

  denom = max(len(gt_boxes), len(pred), 1)
  return 0.2 + 0.8 * (total_iou / denom)
```

```python
box = {'label': 'pothole', 'bbox': [100, 100, 200, 200]}
perfect = json.dumps([{'label': 'pothole', 'bbox_2d': box['bbox']}])
wrong_class = json.dumps([{'label': 'crack', 'bbox_2d': box['bbox']}])
extra = json.dumps([{'label': 'pothole', 'bbox_2d': box['bbox']},
                    {'label': 'pothole', 'bbox_2d': [500, 500, 600, 600]}])
assert compute_reward(perfect, [box]) == 1.0
assert compute_reward('not json', [box]) == 0.0
assert compute_reward(wrong_class, [box]) == 0.2
assert math.isclose(compute_reward(extra, [box]), 0.6)
assert compute_reward('[]', []) == 1.0
assert compute_reward('[]', [box]) == 0.2
assert compute_reward(perfect, []) == 0.2
print('Reward checks: PASS')
```
